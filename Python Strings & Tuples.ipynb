###  String Slicing and Indexing

```python
string4=("Hello Akshaya, welcome to Python programming")
print(string4)

print(string4[0])
print(string4[-1])
print(string4[:5])
print(string4[-11:])
print(string4[::-1])
print(string4[26:32])
```
